## 2. 샘플링
사기(Class=1)는 전부 유지하고, 정상(Class=0)은 10,000건만 무작위 추출해 합친다.

```python
df_fraud = df[df['Class'] == 1]
df_normal = df[df['Class'] == 0].sample(n=10000, random_state=42)

df_sample = pd.concat([df_fraud, df_normal]).reset_index(drop=True)

print('원본 shape:', df.shape)
print('샘플 shape:', df_sample.shape)
print('사기:', len(df_fraud), '/ 정상:', len(df_normal))

print('\n[샘플링 후 Class 비율]')
print(df_sample['Class'].value_counts(normalize=True))
```

## 3. 학습/테스트 분할
스케일링보다 **먼저** 분할한다. Amount는 아직 원본 그대로 두고 X / y로 나눈 뒤,
stratify=y로 클래스 비율을 유지한 채 8:2로 분할한다.

```python
from sklearn.model_selection import train_test_split

# Amount는 원본 그대로 유지한 채 X / y 분리
X = df_sample.drop(columns=['Class'])
y = df_sample['Class']

print('X shape:', X.shape, '/ y shape:', y.shape)
print('X 컬럼:', list(X.columns))

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, stratify=y, random_state=42
)

print('\nX_train:', X_train.shape, '/ X_test:', X_test.shape)

print('\n[y_train Class 비율]')
print(y_train.value_counts(normalize=True))

print('\n[y_test Class 비율]')
print(y_test.value_counts(normalize=True))
```

## 4. 스케일링 (데이터 누수 방지)
StandardScaler는 **X_train에만 fit**하고, X_test에는 그 통계량으로 transform만 적용한다.
두 데이터셋 모두 Amount_Scaled를 만들고 원본 Amount는 제거한다.

```python
from sklearn.preprocessing import StandardScaler

X_train = X_train.copy()
X_test = X_test.copy()

scaler = StandardScaler()
X_train['Amount_Scaled'] = scaler.fit_transform(X_train[['Amount']])  # train에만 fit
X_test['Amount_Scaled'] = scaler.transform(X_test[['Amount']])        # test는 transform만

X_train = X_train.drop(columns=['Amount'])
X_test = X_test.drop(columns=['Amount'])

print('scaler가 학습한 통계량 (train 기준)')
print('  mean:', scaler.mean_[0], '/ std:', scaler.scale_[0])

print('\n[Amount_Scaled 분포]')
print('  X_train  mean: %.6f  std: %.6f' % (X_train['Amount_Scaled'].mean(), X_train['Amount_Scaled'].std()))
print('  X_test   mean: %.6f  std: %.6f' % (X_test['Amount_Scaled'].mean(), X_test['Amount_Scaled'].std()))
print('  -> train은 평균 0 / 표준편차 1, test는 살짝 벗어나는 것이 정상')

print('\nX_train:', X_train.shape, '/ X_test:', X_test.shape)
print('X 컬럼:', list(X_train.columns))
print("Amount 제거 확인:", 'Amount' not in X_train.columns and 'Amount' not in X_test.columns)
```

## 5. SMOTE 오버샘플링
학습 데이터(X_train, y_train)에만 SMOTE를 적용해 소수 클래스를 늘린다.
X_test / y_test는 실제 분포를 그대로 유지해야 하므로 절대 건드리지 않는다.

```python
from imblearn.over_sampling import SMOTE

before_fraud = (y_train == 1).sum()
before_normal = (y_train == 0).sum()

smote = SMOTE(random_state=42)
X_train_smote, y_train_smote = smote.fit_resample(X_train, y_train)  # train에만 적용

after_fraud = (y_train_smote == 1).sum()
after_normal = (y_train_smote == 0).sum()

print('[사기 거래(Class=1) 건수]')
print('  전: %d건 -> 후: %d건 (+%d)' % (before_fraud, after_fraud, after_fraud - before_fraud))

print('\n[정상 거래(Class=0) 건수]')
print('  전: %d건 -> 후: %d건' % (before_normal, after_normal))

print('\n[shape]')
print('  X_train:', X_train.shape, '-> X_train_smote:', X_train_smote.shape)
print('  y_train:', y_train.shape, '-> y_train_smote:', y_train_smote.shape)

print('\n[SMOTE 후 Class 비율]')
print(y_train_smote.value_counts(normalize=True))

print('\n[테스트셋 미변경 확인]')
print('  X_test:', X_test.shape, '/ y_test:', y_test.shape)
print('  y_test 사기 건수:', (y_test == 1).sum(), '(원본 그대로)')
```

## 6. 모델 학습 및 평가 (RandomForest)
SMOTE로 균형을 맞춘 학습 데이터로 RandomForest를 학습하고,
손대지 않은 원본 분포의 X_test로 평가한다. 불균형 데이터이므로
정확도 대신 Precision / Recall / F1과 PR-AUC를 본다.

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report, average_precision_score

# SMOTE로 균형 맞춘 데이터로 학습
model = RandomForestClassifier(random_state=42)
model.fit(X_train_smote, y_train_smote)
print('학습 완료 - 학습 데이터:', X_train_smote.shape)

# 원본 분포의 테스트셋으로 예측
y_pred = model.predict(X_test)
y_proba = model.predict_proba(X_test)[:, 1]  # 사기(Class=1)일 확률

print('\n=== 예측 결과 ===')
print('y_pred  shape:', y_pred.shape, '/ 사기로 예측한 건수:', (y_pred == 1).sum())
print('y_proba shape:', y_proba.shape, '/ 범위: %.4f ~ %.4f' % (y_proba.min(), y_proba.max()))

print('\n=== Classification Report ===')
print(classification_report(y_test, y_pred, target_names=['정상(0)', '사기(1)'], digits=4))

pr_auc = average_precision_score(y_test, y_proba)
print('=== PR-AUC (Average Precision) ===')
print('PR-AUC: %.4f' % pr_auc)
```

## 7. 최종 성능 평가

목표치: Recall ≥ 0.80, F1 ≥ 0.88, PR-AUC ≥ 0.90 (Class 0, 1 모두)

### 클래스별 결과

| 클래스 | Precision | Recall | F1-score |
|---|---|---|---|
| 0 (정상) | 0.9945 | 0.9975 | 0.9960 |
| 1 (사기) | 0.9457 | 0.8878 | 0.9158 |

PR-AUC (Average Precision): **0.9538**

### 목표 대비

| 지표 | 목표 | Class 0 | Class 1 | 달성 |
|---|---|---|---|---|
| Recall | ≥ 0.80 | 0.9975 | 0.8878 | ✅ |
| F1-score | ≥ 0.88 | 0.9960 | 0.9158 | ✅ |
| PR-AUC | ≥ 0.90 | - | 0.9538 | ✅ |

모든 지표가 목표치를 넘었다. 가장 빠듯한 항목은 사기 클래스의 Recall로,
목표 0.80 대비 0.8878을 기록해 0.09 정도 여유가 있다.

### 결론

RandomForest에 SMOTE를 적용한 것만으로 목표를 모두 달성했다.
predict()의 기본 threshold인 0.5를 그대로 사용했고, 별도의 threshold 튜닝이나
class_weight 조정, 하이퍼파라미터 탐색은 하지 않았다.

Recall을 더 끌어올리려면 threshold를 0.5보다 낮추는 방법이 있지만
그만큼 Precision이 떨어진다. 현재는 목표를 충족하므로 추가 조정은 하지 않았다.

참고로 PR-AUC는 사기(Class=1)를 양성으로 두고 계산한 값이다.
`average_precision_score`는 이진 분류에서 양성 클래스 기준 단일 값을 반환하므로
클래스별로 나누어 표기하지 않았다.
